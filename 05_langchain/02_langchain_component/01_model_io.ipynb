# HuggingFace

- 추상화된 채팅LLM: init_chat_model() -> 특정 회사(OpenAI / Anthropoc / HuggingFace) 의 클래스에 직접 묶이지 않고, 공통 인터페이스로 같은 방식(invoke, stream 등)으로 쓰게 만들어준다.

- 즉, init_chat_model() 은 OpenAI 객체를 만드는 방식이 아니라, 여러 채팅 llm 을 공통 규격으로 감싸서 (추상화 하여) 동일하게 쓸 수 있도록 만들어준 팩토리 함수이다.

```python
from langchain.chat_models import init_chat_model # 추상화된 lm 모델 객체 생성

# 객체 생성 : 모델 + 하이퍼파라미터 설정
llm = init_chat_model(
    model = 'huggingface:Qwen/Qwen2.5-0.5B-Instruct', 
    temperature = 1
)

ai_message = llm.invoke('속초 당일치기 여행계획 만들어 줘.') # 프롬프트 -> AIMessage 반환
print(ai_message.content)
```

# 여러 모델 후보군 비교

```python
models = [
    init_chat_model("openai:gpt-5.6-luna"),
    init_chat_model("openai:gpt-4.1-mini"),
    init_chat_model("huggingface:Qwen/Qwen2.5-0.5B-Instruct"),
]

user_prompt = '공부를 잘하는 방법?'

for model in models:
    # model 객체에서 model_name 속성 있으면 model_name, 없으면 model_id 를 사용
    model_name = model.model_name if hasattr(model, 'model_name') else model.model_validate
    print(f'====={model_name}=====')
    ai_message = model.invoke(user_prompt)
    print(ai_message.content)
    print()
```

```python
# PromptTemplate 으로 프롬프트를 템플릿화해서 상품별 문구를 만들고 LLM을 생성

from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    "{product}를 홍보하기 위한 참신한 문구를 생성해 줘."
)
print(prompt)

print(prompt.format(product = '전기차')) # 변수를 채운 완성된 프롬프트 문자열
print(prompt.format(product = '카메라'))

llm = init_chat_model('openai:gpt-5.6-luna')
ai_message = llm.invoke(prompt.format(product = '카메라'))

print(ai_message.content)
```

```python
# PromptTemplate 으로 프롬프트를 템플릿화해서 상품별 문구를 만들고 LLM을 생성

from langchain_core.prompts import ChatPromptTemplate

chat_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "당신은 {domain} 분야의 전문가입니다. 친절히 답해주세요."
    ),
    (
        "user",
        "{question} 한글로 답해주세요."
    ),
])

print(chat_prompt.format(domain='IT', question = 'langsmith 사용법'))

llm = init_chat_model('openai:gpt-5.6-luna')
ai_message = llm.invoke(chat_prompt.format(domain='IT', question = 'langsmith 사용법'))

print(ai_message.content)
```

## Output Parser

객체의 출력에서 실제 값 부분을 가져와 후처리 작업 담당

```python
# CommaSeparatedListOutputParser() : 콤마 구분 문자열 -> 리스트 변환 파서
from langchain_core.output_parsers import CommaSeparatedListOutputParser

parser = CommaSeparatedListOutputParser()
model_output = '달걀, 고구마, 대파, 쪽파'
parser.parse(model_output)
```

```python
format_instruction = parser.get_format_instructions() # 파서가 원하는 출력 형태 지시문 생성
print(format_instruction)
```

```python
prompt = PromptTemplate(
    # subject / n + 지시사항을 포함한 템플릿
    template=(
        "{subject} 팀 {n}개를 작성해주세요.\n"
        "(지시사항: {format_instruction})"
    ),
    input_variables=["subject", "n"], # 사용자가 채울 변수들
    partial_variables={
        "format_instruction": format_instruction
    }, # 고정으로 주입할 변수
)

print(prompt)
print(prompt.format(subject='kbo', n = 7)) # subject = kbo, n = 7을 채운 최종 프롬포트 문자열
```

```python
prompt.invoke({'subject' :'kbo', 'n': 7})
```

```python
# 변수 (subject, n) 을 넣어 프롬프트 메시지 생성
input = prompt.invoke({'subject': 'kbo', 'n': 7}) 
print(input)

llm = init_chat_model('openai:gpt-5.6-luna')
ai_message = llm.invoke(input)
print(ai_message.content)

parser = CommaSeparatedListOutputParser()
output = parser.invoke(ai_message) # LLM 응답을 콤마 기준으로 파싱-> 파이썬 리스트
print(output)
```

```python
# LCEL (파이프)로 Prompt -> LLM -> OutputParser 를 하나의 chain으로 연결해서 실행
chain = prompt | llm | parser

chain.invoke({'subject': 'kbo', 'n' :7})
```

```python
# StrOutputParser : 모델 답변을 가공없이 문자열로 받아서 출력
from langchain_core.output_parsers import StrOutputParser # AIMessage -> str(content) 파서
parser = StrOutputParser()
chain = llm | parser # LLM 출력 -> 문자열로 파싱
chain.invoke('아 너무 덥다 여름에 먹을만한 제철과일은 뭐야?')
```

```python
# JsonOutputParser
from langchain_core.output_parsers import JsonOutputParser
import json

parser = JsonOutputParser()
format_instruction = parser.get_format_instructions()

print(format_instruction)

prompt = PromptTemplate(
    # subject / n + 지시사항을 포함한 템플릿
    template=(
        "{subject} 분야에서 유명한 한국어 책 {n}권 정보 제공해주세요\n"
        "(지시사항: {format_instruction})"
    ),
    input_variables=["subject", "n"], # 사용자가 채울 변수들
    partial_variables={
        "format_instruction": format_instruction
    }, # 고정으로 주입할 변수
)

chain = prompt | llm | parser # 체인 : prompt -> LLM -> JSON 파서
result = chain.invoke({"subject": "인공지능", "n": 3}) # chain 실행
print(json.dumps(result, ensure_ascii=False, indent=2))
```

```python
# PydanticOutputParser : 형식을 강제해서 구조화된 형태로 받는 파서
from pydantic import BaseModel # 데이터 스키마(모델) 정의용
# LLM 출력 -> Pydantic 객체로 검증/변환
from langchain_core.output_parsers import PydanticOutputParser 
from typing import List # 리스트 타입 힌트용

class Book(BaseModel):
    title : str
    author : str
    publisher : str
    year : int = None # 없어도 상관없음
    description : str

# 책 여러권을 담는 스키마 정의
class BookList(BaseModel):
    books : List[Book]

parser = PydanticOutputParser(pydantic_object = BookList)
format_instruction = parser.get_format_instructions()
print(format_instruction)

prompt = PromptTemplate(
    # subject / n + 지시사항을 포함한 템플릿
    template = '{subject} 분야에서 유명한 한국어 책 {n}권 정보를 제공해주세요. \n(지시사항 : {format_instruction})',
    input_variables = ['subject', 'n'], # 사용자가 채울 변수들
    partial_variables = {'format_instruction' : format_instruction} # 고정으로 주입할 변수
)

chain = prompt | llm | parser # 체인 : prompt -> LLM -> JSON 파서
result = chain.invoke({'subject' : '베스트셀러', 'n' : 10}) # chain 실행 
result
```

- Book / Booklist 는 pydantic.BaseModel 을 상속한 스키마(데이터 모델) 이다.
- result 는 그 스키마로 만들어진 Pydantic 객체(인스턴스)
- Pydantic을 사용하면
    - 필드가 있는지 검사
    - 타입 검사 / 변환
    - 구조가 맞는지 검사
- dictionary 와 차이점
    - dict : 아무키나 들어갈 수 있고, 타입도 상관없다.
    - Pydantic 객체 : 정해진 키/타입만 허용. 틀리면 검증 단계에서 수정해줌    

```python
# with_structured_output() : pYDANTIC 스키마를 값으로 받아 해당 형태의 구조화된 결과를 바로 반환
prompt = PromptTemplate(
    # subject / n + 지시사항을 포함한 템플릿
    template = '{subject} 분야에서 유명한 한국어 책 {n}권 정보를 제공해주세요.',
    input_variables = ['subject', 'n'] # 사용자가 채울 변수들
)

chain = prompt | llm.with_structured_output(BookList) # LLM 출력이 BookList 스키마를 따르도록 래핑
result = chain.invoke({'subject' : '인공지능', 'n' : 5}) # chain 실행 
result # 파싱 / 검증된 BookList 객체 반환
```

```python
for book in result.books:
    print(f"책 제목 : {book.title}")
    print(f"책 저자 : {book.author}")
    print(f"출판사 : {book.publisher}")
    print(f"출판연도 : {book.year}")
    print(f"설명 : {book.description}")
    print()
```
